```python
import pandas as pd
import numpy as np
```

```python
#Read data from csv file
data = pd.read_csv("/content/Student_Depression_Dataset.csv")
data.info()
```

```python
data.head()
```

```python
data['Gender'] = data['Gender'].map({'Male':0,'Female':1})
data['Family_background'] = data['Family_background'].map({'Lower Middle class':1,'Middle class':2,'Rich class':3})
```

```python
data.Relationship_status.unique()
```

```python
#Onhot encoding on Relationship_status
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(data[['Relationship_status']])
encoded_df = pd.DataFrame(encoded_data.toarray(), columns=encoder.get_feature_names_out(['Relationship_status']))
data = pd.concat([data, encoded_df], axis=1)
data.drop('Relationship_status', axis=1, inplace=True)
data.head()
```

```python
data.Marital_status.unique()
```

```python
#Onhot encoding on Marital_status
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(data[['Marital_status']])
encoded_df = pd.DataFrame(encoded_data.toarray(), columns=encoder.get_feature_names_out(['Marital_status']))
data = pd.concat([data, encoded_df], axis=1)
data.drop('Marital_status', axis=1, inplace=True)
data.head()
```

```python
data.Health_condition.unique()
```

```python
data['Living_style'] = data['Living_style'].map({'Bachelor':0,'Family':1})
data['friend_circle'] = data['friend_circle'].map({'No':0,'Yes':1})
data['Health_condition'] = data['Health_condition'].map({'Good':3,'Normal':2,'Abnormal':1,'Critical':0})
data['Depression'] = data['Depression'].map({'No':0,'Yes':1})
```

```python
updated_data = data.drop('Home_division', axis=1)
updated_data.head()
```

```python
#Independent var and dependent variable
x = updated_data.drop('Depression', axis=1)
y = updated_data['Depression']
```

```python
x.columns
```

```python
y
```

```python
#Train and test split
from sklearn.model_selection import train_test_split
x_train,y_train,x_test,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Logistic regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
lr = LogisticRegression()
lr.fit(x_train,x_test)
pred = lr.predict(y_train)
accuracy_score(y_test,pred)
```

```python
#Randomforest
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()
rfc.fit(x_train,x_test)
pred = rfc.predict(y_train)
accuracy_score(y_test,pred)
```

```python
#Save as Pickle file
import pickle
pickle.dump(rfc,open('model.pkl','wb'))
```
